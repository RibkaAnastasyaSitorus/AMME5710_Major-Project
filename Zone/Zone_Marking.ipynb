```python
try:
    from google.colab import output
    output.enable_custom_widget_manager()
except ImportError:
    pass
%matplotlib widget
from pathlib import Path
from zone_editor import CameraZoneEditor
from zone_setup import display_file_loader
```

## Open the editor

```python
CONFIG_PATH = Path("zone_config.json").resolve()
editor = CameraZoneEditor(str(CONFIG_PATH))
display_file_loader(editor, folder=str(Path.cwd()))
print("Zone configuration:", CONFIG_PATH)
```

## Save your configuration outside the Colab runtime
After marking and saving zones, run this cell to download the JSON.


```python
if not CONFIG_PATH.is_file():
    print("Save a camera and its zones first.")
else:
    try:
        from google.colab import files
        files.download(str(CONFIG_PATH))
    except ImportError:
        print("Configuration saved locally:", CONFIG_PATH)
```

## Show saved zone IDs for integration
Use these IDs when exporting workstation/walkway polygons.


```python
from zone_core import load_config
if CONFIG_PATH.is_file():
    config = load_config(str(CONFIG_PATH))
    for camera_id, camera in config["cameras"].items():
        for zone_id in camera["zones"]:
            rule = config["zone_rules"][zone_id]
            print(camera_id, zone_id, rule["name"], rule["required_ppe"])
```
